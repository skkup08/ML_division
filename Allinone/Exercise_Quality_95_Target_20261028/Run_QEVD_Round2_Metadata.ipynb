```python
import subprocess, pathlib, sys, datetime
root = pathlib.Path('/content/drive/MyDrive/Allinone/Exercise_Quality_95_Target_20261028')
script = root / 'resolve_qevd_round2_metadata.py'
log = root / 'qevd_round2_metadata_execution.log'
if not script.exists():
    raise FileNotFoundError(script)
proc = subprocess.run([sys.executable, str(script)], text=True, capture_output=True)
payload = f'UTC={datetime.datetime.now(datetime.timezone.utc).isoformat()}\nRETURN_CODE={proc.returncode}\nSTDOUT\n{proc.stdout}\nSTDERR\n{proc.stderr}'
log.write_text(payload)
print(proc.stdout)
if proc.stderr:
    print(proc.stderr)
if proc.returncode != 0:
    raise RuntimeError(f'metadata resolver failed: {proc.returncode}')
print('ROUND2_METADATA_NOTEBOOK_COMPLETE')
```
